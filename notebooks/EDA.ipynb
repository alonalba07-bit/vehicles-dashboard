```python
# Librerias
import pandas as pd
import plotly.express as px

# Dataframe
vehicles = pd.read_csv('C:\Users\Asus ROG\OneDrive - Universidad Autónoma del Estado de México\Documents\Personnelle\CV\Bootcamp\Triple_Tent\Cientifico_datos\Sprint_7\Proyecto\vehicles-dashboard')

#Inicio
vehicles.head()
vehicles.info()
vehicles.describe()
vehicles.isna().sum()

# Histograma Plotly
fig = px.histogram(
    vehicles,
    x='odometer',
    title='Distribución del kilometraje de los vehículos'
)

fig.show()

# Dispersion
fig = px.scatter(
    vehicles,
    x='odometer',
    y='price',
    title='Relación entre kilometraje y precio de los vehículos'
)

fig.show()
```
